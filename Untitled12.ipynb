```python
# prompt: call a numpy function from tensorflow that returns a 1-D tensor  tf.numpy_function

import tensorflow as tf
import tensorflow_datasets as tfds
import numpy as np
import tensorflow_io as tfio


def pad_to_multiple_of_1024(tensor):
    pad_size = 1024 - (tf.shape(tensor)[0] % 1024)
    return tf.cast(tf.pad(tensor, [[0, pad_size]]), tf.float32)

def apply_spectrogram(tensor):
    return tfio.audio.spectrogram(tensor, 256, 256, 64, name=None)

def apply_gl(tensor):
    return tfio.audio.inverse_spectrogram(tensor, 256, 256, 64, iterations=30)

def split_sizes_calc(num_blocks):
    return np.array([16] * num_blocks, dtype=np.int32)

def split_into_blocks(tensor, block_size=1024):
    tensor_length = tf.shape(tensor)[0]
    num_blocks = tensor_length // (block_size // 64)
    split_sizes = tf.numpy_function(split_sizes_calc, [num_blocks], tf.int32)
    split_sizes = tf.reshape(split_sizes, [-1])
    splits = tf.split(tensor, num_or_size_splits=split_sizes, axis=0)
    return tf.stack(splits)

def group_back(tensor):
    original_shape = tf.shape(tensor)
    return tf.reshape(tensor, [-1, original_shape[2]])

def process_dataset(dataset):
    dataset = dataset.map(lambda x: x['audio'])
    dataset = dataset.map(lambda x: pad_to_multiple_of_1024(x))
    dataset = dataset.map(lambda x: apply_spectrogram(x))
    dataset = dataset.map(lambda x: split_into_blocks(x))
    return dataset

train_dataset = tfds.load('spoken_digit', split='train', as_supervised=False)
processed_dataset = process_dataset(train_dataset)
tf.print(processed_dataset)
```
