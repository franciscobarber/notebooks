````python
import tensorflow as tf
import tensorflow_datasets as tfds
import numpy as np
import tensorflow_io as tfio
import torch
import librosa
def pad_to_multiple_of_1024(tensor):
    pad_size = 1024 - (tf.shape(tensor)[0] % 1024)
    return tf.cast(tf.pad(tensor, [[0, pad_size]]), tf.float32)

def apply_spectrogram(tensor):
    spectrogram = tfio.audio.spectrogram(tensor, 256, 256, 64, name=None)
    #log_spectrogram = tf.math.log(spectrogram + 1e-6)
    # Clip values to a specific range (e.g., -10 to 10)
    # Equivalent of librosa.power_to_db
    log_ref = 5e-0
    power_to_db = 10.0 * tf.math.log(spectrogram + 1e-6) / tf.math.log(10.0)
    power_to_db = power_to_db - 10.0 * tf.math.log(log_ref) / tf.math.log(10.0)
    #power_to_db = librosa.power_to_db(spectrogram,ref=log_ref)
    #clipped_spectrogram = tf.clip_by_value(power_to_db, 0.0, 1.0)
    #return clipped_spectrogram
    return power_to_db
def calculate_spectrogram_with_librosa(audio):
    """Calculates spectrogram using Torch and converts to power-to-dB using librosa."""

    # Use tf.py_function to execute NumPy/Torch code within TensorFlow graph
    spectrogram_tf = tf.py_function(
        _calculate_spectrogram_with_librosa_numpy,
        [audio],  # Input tensor to tf.py_function
        tf.float32  # Output data type
    )

    return spectrogram_tf

def _calculate_spectrogram_with_librosa_numpy(audio):
    """Helper function to perform NumPy/Torch operations."""

    # Convert TensorFlow tensor to NumPy array
    audio_np = audio.numpy()

    # Calculate spectrogram using Torch
    y = torch.from_numpy(audio_np).float()
    windowsize = 256
    window = torch.hann_window(windowsize)
    S = torch.stft(y, windowsize, window=window, return_complex=False)

    # Discard phase information
    mag = S.pow(2).sum(2).sqrt()
    ms = mag.numpy()

    # Define log_ref for librosa.power_to_db
    log_ref = np.max(ms)
    spectrogram_db = librosa.power_to_db(ms, ref=log_ref)
    spectrogram_db = spectrogram_db.T

    return tf.convert_to_tensor(spectrogram_db, dtype=tf.float32)
def apply_gl(tensor):
    log_ref = 5e-0
    tensor = tf.pow(10.0, (tensor / 10.0)) * log_ref
    return tfio.audio.inverse_spectrogram(tensor, 256, 256, 64, iterations=30)
def split_sizes_calc(num_blocks):
    split_sizes = [16] * num_blocks
    return split_sizes
def split_into_blocks(tensor, block_size=1024):
    tensor_length = tf.shape(tensor)[0]
    block_length = block_size // 64
    num_blocks = tensor_length // block_length
    splits = tf.TensorArray(dtype=tensor.dtype, size=num_blocks)
    i = tf.constant(0)

    def condition(i, splits):
        return i < num_blocks

    def body(i, splits):
        start = i * block_length
        end = (i + 1) * block_length
        block = tensor[start:end]
        splits = splits.write(i, block)
        return i + 1, splits

    _, splits = tf.while_loop(condition, body, [i, splits])

    return splits.stack()

def flatten(tensor):
    """Flattens a tensor into a 1-D tensor."""
    return tf.reshape(tensor, [-1])

#def split_into_blocks_np(tensor, block_size=1024):
#    tensor_length = tensor.shape[0]
#    num_blocks = tensor_length // (block_size // 64)
#    split_size = block_size // 64
#    splits = np.array_split(tensor, num_blocks)
#    return splits
#def split_into_blocks(tensor, block_size=1024):
#    splits = tf.numpy_function(split_into_blocks_np, [tensor, block_size], [tf.float32])
#    return tf.data.Dataset.from_tensor_slices(splits)
def group_back(tensor):
    original_shape = tf.shape(tensor)
    return tf.reshape(tensor, [-1, original_shape[2]])

def process_dataset(dataset, max_value=None, min_value=None):
    dataset = dataset.map(lambda x: x['audio'])
    dataset = dataset.map(lambda x: pad_to_multiple_of_1024(x))
    dataset = dataset.map(lambda x: apply_spectrogram(x))

    # Calculate min/max if not provided
    if max_value is None or min_value is None:
        min_value = dataset.reduce(
            tf.constant(np.inf, dtype=tf.float32),
            lambda min_so_far, spectrogram: tf.math.minimum(min_so_far, tf.math.reduce_min(spectrogram))
        )
        max_value = dataset.reduce(
            tf.constant(-np.inf, dtype=tf.float32),
            lambda max_so_far, spectrogram: tf.math.maximum(max_so_far, tf.math.reduce_max(spectrogram))
        )

    # Clip and scale spectrograms
    epsilon = 1e-6
    dataset = dataset.map(lambda x: (x - min_value + epsilon) / (max_value - min_value + 2 * epsilon))

    dataset = dataset.flat_map(lambda x: tf.data.Dataset.from_tensor_slices(split_into_blocks(x)))
    dataset = dataset.map(lambda x: flatten(x))
    dataset = dataset.map(lambda x: (x, x))

    return dataset, max_value, min_value

# Processing the datasets
train_dataset, val_dataset, test_dataset = tfds.load(
    'spoken_digit', split=['train[:70%]', 'train[70%:80%]', 'train[80%:]'], as_supervised=False
)

# Process training dataset and get min/max values
processed_dataset, max_value, min_value = process_dataset(train_dataset)

# Process validation and test datasets using training min/max values
processed_val_dataset, _, _ = process_dataset(val_dataset, max_value, min_value)
processed_test_dataset, _, _ = process_dataset(test_dataset, max_value, min_value)
````

````python
import tensorflow as tf
import tensorflow_datasets as tfds
import numpy as np
import tensorflow_io as tfio
import torch
import librosa
def pad_to_multiple_of_1024(tensor):
    pad_size = 1024 - (tf.shape(tensor)[0] % 1024)
    return tf.cast(tf.pad(tensor, [[0, pad_size]]), tf.float32)

def apply_dct(audio):
    dct_coeff = tf.signal.dct(tf.cast(audio, tf.float32))
    log_ref = 5e-0
    power_to_db = 10.0 * tf.math.log1p(tf.abs(dct_coeff)) / tf.math.log(10.0)
    power_to_db = power_to_db - 10.0 * tf.math.log(log_ref) / tf.math.log(10.0)
    # Clip to a safe range to avoid extremely large values
    power_to_db = tf.clip_by_value(power_to_db, -100.0, 100.0)
    return power_to_db

def apply_idct(tensor):
    log_ref = 5e-0
    tensor = tf.pow(10.0, (tensor / 10.0)) * log_ref
    return tf.signal.idct(tensor)

def split_into_blocks(tensor, block_size=1024):
    tensor_length = tf.shape(tensor)[0]
    block_length = block_size
    num_blocks = tensor_length // block_length
    splits = tf.TensorArray(dtype=tensor.dtype, size=num_blocks)
    i = tf.constant(0)

    def condition(i, splits):
        return i < num_blocks

    def body(i, splits):
        start = i * block_length
        end = (i + 1) * block_length
        block = tensor[start:end]
        splits = splits.write(i, block)
        return i + 1, splits

    _, splits = tf.while_loop(condition, body, [i, splits])

    return splits.stack()


def group_back(tensor):
    original_shape = tf.shape(tensor)
    return tf.reshape(tensor, [-1, original_shape[2]])

def process_dataset(dataset, max_value=None, min_value=None):
    dataset = dataset.map(lambda x: x['audio'])
    dataset = dataset.map(lambda x: pad_to_multiple_of_1024(x))

    dataset = dataset.flat_map(lambda x: tf.data.Dataset.from_tensor_slices(split_into_blocks(x)))
    dataset = dataset.map(lambda x: apply_dct(x))

    # Calculate min/max if not provided
    if max_value is None or min_value is None:
        min_value = dataset.reduce(
            tf.constant(np.inf, dtype=tf.float32),
            lambda min_so_far, spectrogram: tf.math.minimum(min_so_far, tf.math.reduce_min(spectrogram))
        )
        max_value = dataset.reduce(
            tf.constant(-np.inf, dtype=tf.float32),
            lambda max_so_far, spectrogram: tf.math.maximum(max_so_far, tf.math.reduce_max(spectrogram))
        )

    # Clip and scale spectrograms
    epsilon = 1e-6
    dataset = dataset.map(lambda x: (x - min_value + epsilon) / (max_value - min_value + 2 * epsilon))

    dataset = dataset.map(lambda x: (x, x))

    return dataset, max_value, min_value

# Processing the datasets
train_dataset, val_dataset, test_dataset = tfds.load(
    'spoken_digit', split=['train[:70%]', 'train[70%:80%]', 'train[80%:]'], as_supervised=False
)

# Process training dataset and get min/max values
processed_dataset, max_value, min_value = process_dataset(train_dataset)

# Process validation and test datasets using training min/max values
processed_val_dataset, _, _ = process_dataset(val_dataset, max_value, min_value)
processed_test_dataset, _, _ = process_dataset(test_dataset, max_value, min_value)
batched_train_dataset = processed_dataset.batch(64)
batched_val_dataset = processed_val_dataset.batch(64)

FRAME_LENGTH=1024
input_img = Input(shape=(FRAME_LENGTH,))
encoded = Dense(256)(input_img)
decoded = Dense(FRAME_LENGTH, activation='sigmoid')(encoded)
autoencoder = Model(input_img, decoded)
autoencoder.compile(optimizer='adam', loss='mse')


# Train the autoencoder
with tf.device('/GPU:0'):
    autoencoder.fit(batched_train_dataset,
                    epochs=50,
                    batch_size=64,
                    shuffle=True,
                    validation_data=batched_val_dataset)
````
